```python
!pip install -q pypdf

import pypdf

def extract_text_from_pdf(pdf_path):
    reader = pypdf.PdfReader(pdf_path)
    extracted_text = []
    for page_num, page in enumerate(reader.pages):
        text = page.extract_text()
        extracted_text.append(f"--- Page {page_num + 1} ---\n{text}")
    return "\n".join(extracted_text)

pdf_file_path = "/content/Bharath_Naukri_LinkedIn_AI_Profile_Guide.pdf"
try:
    text_content = extract_text_from_pdf(pdf_file_path)
    print("PDF content successfully extracted:\n")
    print(text_content[:2000] + "\n... [Truncated if too long] ...")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
from sentence_transformers import SentenceTransformer
model=SentenceTransformer("all-MiniLM-L6-v2")
text=text_content.splitlines()
embeddings=model.encode(text)
print(embeddings.shape)
```

```python
vectors=[]
for i,linetext in enumerate(text):
  vectors.append({"id":str(i),
                 "values":embeddings[i].tolist(),
                  "metadata":{
                      "text":linetext,
                      "category":"Analyst"
                  }})
```

```python
vectors[0:1]
```

```python
index.upsert(vectors=vectors)
```

```python
query="what is data anlytics"
query_embedding=model.encode(query).tolist()
results=index.query(vector=query_embedding,top_k=3,include_metadata=True)
```

```python
for i in results.matches:
  print(i.id)
  print(i.score)
  print(i.metadata['text'])
```
