```python
#Q1. Write a function for line total for banking case 1; use sample reference ACC-1001 and explain the result.


from functools import reduce

Acc_num = "ACC-1001"

transactions = [
    {"account": "ACC-1001", "quantity": 5, "price": 200},
    {"account": "ACC-1002", "quantity": 3, "price": 150},
    {"account": "ACC-1003", "quantity": 0, "price": 300}
]

# Pure Function
def line_total(record):
    return record["quantity"] * record["price"]

# Select valid transactions
valid_transactions = list(filter(lambda record: record["quantity"] > 0,transactions))

# Transform transactions into line totals
totals = list(map(line_total,valid_transactions))

# Calculate grand total
grand_total = reduce(lambda x, y: x + y,totals,0)

print("Valid Transactions:", valid_transactions)
print("Line Totals:", totals)
print(f"{Acc_num} Grand Total:", grand_total)
```

```python
#Q39. Write a generator for low-balance accounts for banking case 1; use sample reference ACC-1039 and explain the result.


accounts = [
    ("ACC-1001", 25000),
    ("ACC-1002", 5000),
    ("ACC-1003", 8000)
]

def low_balance_accounts(accounts):
    for account, balance in accounts:
        if balance < 10000:
            yield account, balance

for account, balance in low_balance_accounts(accounts):
    print(account, balance)

# a generator is used to produce low-balance accounts one at a time using the yield keyword.
# The generator checks each account balance and yields only balances below ₹10,000.
# This approach is memory efficient because values are generated on demand instead of storing all results in memory at once.
```

```python
#Q41. Validate arguments with a wrapper for banking case 1; use sample reference ACC-1041 and explain the result.


Acc_num = "ACC-1041"

def validate_arguments(func):

    def wrapper(account, amount):

        if not account.startswith("ACC-"):
            return "Invalid account reference"

        if amount <= 0:
            return "Amount must be greater than 0"

        return func(account, amount)

    return wrapper


@validate_arguments
def process_transaction(account, amount):
    return f"{account} Transaction Amount = {amount}"


result = process_transaction(Acc_num, 5000)

print(result)

# a wrapper validates the account reference and transaction amount before calling the original function.
# If the account does not begin with ACC- or the amount is not positive, the wrapper returns an error message.
# If both arguments are valid, it calls the original transaction function. This keeps validation separate from the main banking logic.
```

```python
#Q44. Calculate basket metrics in one pass for banking case 1; use sample reference ACC-1044 and explain the result.


Acc_num = "ACC-1044"

basket = [1000, 2000, 1500, 500]

def basket_metrics(amounts):
    count = 0
    total = 0
    minimum = None
    maximum = None

    for amount in amounts:
        count += 1
        total += amount

        if minimum is None or amount < minimum:
            minimum = amount

        if maximum is None or amount > maximum:
            maximum = amount

    average = total / count if count > 0 else 0

    return count, total, minimum, maximum, average


count, total, minimum, maximum, average = basket_metrics(basket)

print(f"{Acc_num} Basket Report")
print("Transaction Count =", count)
print("Total Amount =", total)
print("Minimum Amount =", minimum)
print("Maximum Amount =", maximum)
print("Average Amount =", average)

# the basket metrics are calculated in one pass using a single loop. During each iteration,
# the function updates the transaction count, total, minimum, and maximum. After the loop, it calculates the average from the total and count.
# For the basket [1000, 2000, 1500, 500], the count is 4, total is 5000, minimum is 500, maximum is 2000, and average is 1250.0.
```

```python
#Q78. Implement retry as a higher-transfer function for banking case 1; use sample reference ACC-1078 and explain the result.


Acc_num = "ACC-1078"


def retry(func, max_attempts=3):
    def wrapper(*args, **kwargs):
        for attempt in range(1, max_attempts + 1):

            try:
                print(f"Attempt {attempt}")
                return func(*args, **kwargs)

            except ConnectionError as error:
                print("Temporary failure:", error)

                if attempt == max_attempts:
                    return "Transfer failed after all retry attempts"

    return wrapper

attempt_count = 0


def transfer_amount(account, amount):
    global attempt_count
    attempt_count += 1

    if attempt_count < 3:
        raise ConnectionError("Banking service unavailable")

    return f"{account} Transfer successful: {amount}"


safe_transfer = retry(transfer_amount, max_attempts=3)

result = safe_transfer(Acc_num, 5000)

print(result)

# retry() is a higher-order function because it accepts a transfer function and returns a wrapper.
# The wrapper calls the original function inside a limited retry loop. Temporary ConnectionError failures are caught, and the operation is retried up to three times.
# In this example, the first two attempts fail and the third returns a successful transfer result.
```

```python
#Q80. Design an end-to-end transactions analytics pipeline for banking case 1; use sample reference ACC-1080 and explain the result.

#Below is the follow: which I followed

#Raw Transactions -> Validation -> Filter Successful Transactions -> Apply Processing Charge -> Group by Category -> Calculate Analytics Report


Acc_num = "ACC-1080"

transactions = [
    {"account": "ACC-1080","category": "Deposit","amount": 5000,"status": "Success"},
    {"account": "ACC-1080","category": "Withdrawal","amount": 2000,"status": "Success"},
    {"account": "ACC-1080","category": "Deposit","amount": 3000,"status": "Failed"},
    {"account": "ACC-1080","category": "Charges","amount": 500,"status": "Success"}
]


# Step 1: Validate one transaction
def is_valid(transaction):
    return (
        transaction.get("account") == "ACC-1080"
        and transaction.get("category")
        and isinstance(transaction.get("amount"), (int, float))
        and transaction.get("amount") > 0
        and transaction.get("status") in {"Success", "Failed"}
    )


# Step 2: Apply a 2% processing charge
def add_processing_charge(transaction):
    processed_amount = transaction["amount"] * 1.02

    return {**transaction,"processed_amount": processed_amount}


# Step 3: Calculate analytics in one pass
def calculate_metrics(transactions):
    count = 0
    total = 0
    minimum = None
    maximum = None

    for transaction in transactions:
        amount = transaction["processed_amount"]

        count += 1
        total += amount

        if minimum is None or amount < minimum:
            minimum = amount

        if maximum is None or amount > maximum:
            maximum = amount

    average = total / count if count > 0 else 0

    return {
        "count": count,
        "total": total,
        "minimum": minimum,
        "maximum": maximum,
        "average": average
    }


# Step 4: Run the complete pipeline
def run_analytics_pipeline(transactions):

    valid_transactions = list(filter(is_valid, transactions))

    successful_transactions = list(filter(lambda transaction:transaction["status"] == "Success",valid_transactions))

    transformed_transactions = list(map(add_processing_charge,successful_transactions))

    grouped_transactions = {}

    for transaction in transformed_transactions:
        category = transaction["category"]

        grouped_transactions.setdefault(category,[]).append(transaction)

    grouped_report = {
        category: calculate_metrics(category_transactions)
        for category, category_transactions
        in grouped_transactions.items()
    }

    overall_report = calculate_metrics(
        transformed_transactions
    )

    return {
        "reference": Acc_num,
        "valid_count": len(valid_transactions),
        "successful_count": len(successful_transactions),
        "grouped_report": grouped_report,
        "overall_report": overall_report
    }


report = run_analytics_pipeline(transactions)


print(report["reference"], "Transactions Analytics Report")
print("Valid Transactions =", report["valid_count"])
print("Successful Transactions =", report["successful_count"])

print("\nCategory Report")

for category, metrics in report["grouped_report"].items():
    print("\nCategory =", category)
    print("Count =", metrics["count"])
    print("Total =", metrics["total"])
    print("Minimum =", metrics["minimum"])
    print("Maximum =", metrics["maximum"])
    print("Average =", metrics["average"])

print("\nOverall Report")

for metric, value in report["overall_report"].items():
    print(metric.capitalize(), "=", value)
```

```python
#Q88. Implement retry as a higher-transfer function for banking case 2; use sample reference ACC-1088 and explain the result.


Acc_num = "ACC-1088"


def retry(func, max_attempts=3):
    def wrapper(*args, **kwargs):
        for attempt in range(1, max_attempts + 1):

            try:
                print(f"Attempt {attempt}")
                return func(*args, **kwargs)

            except ConnectionError as error:
                print("Temporary failure:", error)

                if attempt == max_attempts:
                    return "Transfer failed after all retry attempts"

    return wrapper

attempt_count = 0


def transfer_amount(account, amount):
    global attempt_count
    attempt_count += 1

    if attempt_count < 3:
        raise ConnectionError("Banking service unavailable")

    return f"{account} Transfer successful: {amount}"


safe_transfer = retry(transfer_amount, max_attempts=3)

result = safe_transfer(Acc_num, 5000)

print(result)

# retry() is a higher-order function because it accepts a transfer function and returns a wrapper.
# The wrapper calls the original function inside a limited retry loop. Temporary ConnectionError failures are caught, and the operation is retried up to three times.
# In this example, the first two attempts fail and the third returns a successful transfer result.
```
