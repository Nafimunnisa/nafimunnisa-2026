## Assignment Operators Demonstration

```python
# Initial assignment
x = 10
print(f"Initial value of x: {x}")

# += (Add and assign)
x += 5 # Equivalent to x = x + 5
print(f"After x += 5: {x}")

# -= (Subtract and assign)
x = 10 # Reset x for clarity
x -= 3 # Equivalent to x = x - 3
print(f"After x -= 3: {x}")

# *= (Multiply and assign)
x = 10 # Reset x
x *= 2 # Equivalent to x = x * 2
print(f"After x *= 2: {x}")

# /= (Divide and assign)
x = 10 # Reset x
x /= 2 # Equivalent to x = x / 2
print(f"After x /= 2: {x}")

# //= (Floor divide and assign)
x = 10 # Reset x
x //= 3 # Equivalent to x = x // 3
print(f"After x //= 3: {x}")

# %= (Modulus and assign)
x = 10 # Reset x
x %= 3 # Equivalent to x = x % 3
print(f"After x %= 3: {x}")

# **= (Exponentiate and assign)
x = 2 # Reset x for this operation
x **= 3 # Equivalent to x = x ** 3
print(f"After x **= 3: {x}")
```

## Bank Account Balance Update using Compound Assignment Operators

```python
# Initial bank balance
balance = 1000.0
print(f"Initial balance: ${balance:.2f}")

# Annual interest rate (e.g., 5%)
interest_rate = 0.05

# Simulate interest for one year using compound assignment
# balance += (balance * interest_rate) can be written as balance *= (1 + interest_rate)
balance *= (1 + interest_rate)
print(f"Balance after 1 year (with 5% interest): ${balance:.2f}")

# Simulate another year of interest
balance *= (1 + interest_rate)
print(f"Balance after 2 years: ${balance:.2f}")

# Let's say a deposit is made
deposit = 200.0
balance += deposit # Using += for deposit
print(f"Balance after deposit of ${deposit:.2f}: ${balance:.2f}")

# Let's say a withdrawal is made
withdrawal = 50.0
balance -= withdrawal # Using -= for withdrawal
print(f"Balance after withdrawal of ${withdrawal:.2f}: ${balance:.2f}")
```

## Bitwise Shift Operations

Bitwise shift operators move the bits of a number to the left or right, effectively multiplying or dividing the number by powers of 2.

*   `<<` (Left Shift): Shifts the bits of the number to the left by a specified number of positions. This is equivalent to multiplying the number by 2 raised to the power of the shift amount. New bits on the right are filled with zeros.
*   `>>` (Right Shift): Shifts the bits of the number to the right by a specified number of positions. This is equivalent to integer division of the number by 2 raised to the power of the shift amount. New bits on the left are filled with zeros for positive numbers (arithmetic right shift for signed integers).

```python
# Define a number for bitwise shift operations
number = 10  # Binary: 1010

print(f"Original Number: {number} (Binary: {bin(number)})\n")

# Left Shift (<<)
# Shifting 10 (1010) left by 1 position becomes 20 (10100)
left_shift_1 = number << 1
print(f"Left Shift by 1 ({number} << 1): {left_shift_1} (Binary: {bin(left_shift_1)})")

# Shifting 10 (1010) left by 2 positions becomes 40 (101000)
left_shift_2 = number << 2
print(f"Left Shift by 2 ({number} << 2): {left_shift_2} (Binary: {bin(left_shift_2)})")


# Right Shift (>>)
# Shifting 10 (1010) right by 1 position becomes 5 (0101)
right_shift_1 = number >> 1
print(f"\nRight Shift by 1 ({number} >> 1): {right_shift_1} (Binary: {bin(right_shift_1)})")

# Shifting 10 (1010) right by 2 positions becomes 2 (0010)
right_shift_2 = number >> 2
print(f"Right Shift by 2 ({number} >> 2): {right_shift_2} (Binary: {bin(right_shift_2)})")

# --- Example with a different number ---
number_2 = 16 # Binary: 10000
print(f"\nOriginal Number 2: {number_2} (Binary: {bin(number_2)})\n")

# Left Shift
left_shift_3 = number_2 << 1
print(f"Left Shift by 1 ({number_2} << 1): {left_shift_3} (Binary: {bin(left_shift_3)})")

# Right Shift
right_shift_3 = number_2 >> 2
print(f"Right Shift by 2 ({number_2} >> 2): {right_shift_3} (Binary: {bin(right_shift_3)})")
```

```python
def count_set_bits(n):
    """Counts the number of set bits (1s) in the binary representation of n."""
    count = 0
    current_number = n

    print(f"Original number: {n} (Binary: {bin(n)})")
    print("--- Counting Set Bits ---")

    while current_number > 0:
        # Clear the rightmost set bit
        current_number = current_number & (current_number - 1)
        count += 1
        print(f"  After operation, number becomes: {current_number} (Binary: {bin(current_number)}), count = {count}")

    return count

# Test cases
number1 = 12  # Binary: 1100 (2 set bits)
print(f"\nTotal set bits in {number1}: {count_set_bits(number1)}")

number2 = 7   # Binary: 0111 (3 set bits)
print(f"\nTotal set bits in {number2}: {count_set_bits(number2)}")
```

```python
# ---  Int - Same values, same objects (due to optimization for small integers) ---
a = 10
b = 10

print(f"a = {a}, b = {b}")
print(f"a == b: {a == b}") # Checks if values are equal
print(f"a is b: {a is b}")   # Checks if they are the same object
print(f"id(a): {id(a)}, id(b): {id(b)}\n") # Check memory addresses

# --- List- Same values, different objects (e.g., lists) ---
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(f"list1 = {list1}, list2 = {list2}")
print(f"list1 == list2: {list1 == list2}") # Values are equal
print(f"list1 is list2: {list1 is list2}")   # Different objects in memory
print(f"id(list1): {id(list1)}, id(list2): {id(list2)}\n")

# --- Strings (often optimized for common or short strings) ---
str1 = "hello"
str2 = "hello"
str3 = "h" + "ello"

print(f"str1 = '{str1}', str2 = '{str2}', str3 = '{str3}'")
print(f"str1 == str2: {str1 == str2}")
print(f"str1 is str2: {str1 is str2}") # Often True for short/interned strings
print(f"str1 == str3: {str1 == str3}")
print(f"str1 is str3: {str1 is str3}") # Often True if Python interns 'hello'
print(f"id(str1): {id(str1)}, id(str2): {id(str2)}, id(str3): {id(str3)}\n")

# --- Example 5: `is not` operator ---
x = 5
y = 6
z = 5

print(f"x = {x}, y = {y}, z = {z}")
print(f"x is not y: {x is not y}")
print(f"x is not z: {x is not z}") # False because x and z are the same object
```

```python
# Create two independent list objects with the same content
list_a = [10, 20, 30]
list_b = [10, 20, 30]

print(f"List A: {list_a}, ID: {id(list_a)}")
print(f"List B: {list_b}, ID: {id(list_b)}")
print(f"Are list_a and list_b the same object? (list_a is list_b): {list_a is list_b}")
print(f"Do list_a and list_b have the same values? (list_a == list_b): {list_a == list_b}")

# Now, create a third variable that references the same list object as list_a
list_c = list_a

print(f"\nList A: {list_a}, ID: {id(list_a)}")
print(f"List C: {list_c}, ID: {id(list_c)}")
print(f"Are list_a and list_c the same object? (list_a is list_c): {list_a is list_c}")
print(f"Do list_a and list_c have the same values? (list_a == list_c): {list_a == list_c}")

# Demonstrate what happens when you modify one if they are the same object
list_c.append(40)
print(f"\nAfter modifying list_c:")
print(f"List A: {list_a}, ID: {id(list_a)}")
print(f"List C: {list_c}, ID: {id(list_c)}")
print(f"Are list_a and list_c the same object? (list_a is list_c): {list_a is list_c}")
print(f"Do list_a and list_c have the same values? (list_a == list_c): {list_a == list_c}")
```

```python
# --- `is not` with Integers ---
# Python often interns small integers, making `is` sometimes behave like `==`.
# However, for larger integers or explicitly different assignments, `is not` works as expected.

int_a = 5
int_b = 6
int_c = 5 # Often points to the same object as int_a due to interning
int_d = 257 # Larger integer, often not interned, creating a new object
int_e = 257 # May or may not be the same object as int_d depending on runtime/version

print(f"--- Integers ---")
print(f"int_a = {int_a}, int_b = {int_b}, int_c = {int_c}, int_d = {int_d}, int_e = {int_e}")
print(f"id(int_a): {id(int_a)}, id(int_b): {id(int_b)}, id(int_c): {id(int_c)}")
print(f"id(int_d): {id(int_d)}, id(int_e): {id(int_e)}\n")

print(f"int_a is not int_b: {int_a is not int_b}") # True, different values, different objects
print(f"int_a is not int_c: {int_a is not int_c}") # False, often same object due to interning
print(f"int_d is not int_e: {int_d is not int_e}") # Can be True or False, depends on interning for large integers

# --- `is not` with Lists ---
# Lists are mutable, so even with identical content, they are usually different objects in memory

list_x = [10, 20]
list_y = [10, 20]
list_z = list_x # list_z refers to the *same* object as list_x

print(f"\n--- Lists ---")
print(f"list_x = {list_x}, ID: {id(list_x)}")
print(f"list_y = {list_y}, ID: {id(list_y)}")
print(f"list_z = {list_z}, ID: {id(list_z)}\n")

print(f"list_x is not list_y: {list_x is not list_y}") # True, usually different objects
print(f"list_x is not list_z: {list_x is not list_z}") # False, they are the same object

# Demonstrate modification effect
list_x.append(30)
print(f"\nAfter list_x.append(30):")
print(f"list_x: {list_x}")
print(f"list_z: {list_z}") # list_z also changes because it's the same object
print(f"list_x is not list_z: {list_x is not list_z}")
```

```python
# Define the string to analyze
input_string = "Python Programming Is Fun!"

# Define a list of vowels (both lowercase and uppercase for case-insensitivity)
vowels = ['a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U']

vowel_count = 0

print(f"Analyzing the string: '{input_string}'\n")

# Iterate through each character in the string
for char in input_string:
    # Check if the character is in the list of vowels
    if char in vowels:
        vowel_count += 1
        print(f"Found vowel: '{char}' (Current count: {vowel_count})")

print(f"\nTotal number of vowels found: {vowel_count}")

# Alternative: Convert the string to lowercase first for simpler vowel list
input_string_lower = input_string.lower()
vowels_lower = ['a', 'e', 'i', 'o', 'u']
case_insensitive_vowel_count = 0

print(f"\n--- Case-insensitive counting ---")
print(f"Analyzing the lowercase string: '{input_string_lower}'\n")

for char in input_string_lower:
    if char in vowels_lower:
        case_insensitive_vowel_count += 1
        # Optional: print(f"Found vowel: '{char}' (Current count: {case_insensitive_vowel_count})")

print(f"Total number of vowels (case-insensitive): {case_insensitive_vowel_count}")
```

```python
# Get input from the user
# Using float to allow for decimal numbers
num1 = 12
operation = "+"
num2 = 17

result = None

# Use comparison operators to determine the operation
if operation == '+':
    result = num1 + num2
elif operation == '-':
    result = num1 - num2
elif operation == '*':
    result = num1 * num2
elif operation == '/':
    if num2 == 0:
        result = "Error: Division by zero is not allowed."
    else:
        result = num1 / num2
elif operation == '//':
    if num2 == 0:
        result = "Error: Floor division by zero is not allowed."
    else:
        result = num1 // num2
elif operation == '%':
    if num2 == 0:
        result = "Error: Modulus by zero is not allowed."
    else:
        result = num1 % num2
elif operation == '**':
    result = num1 ** num2
else:
    result = "Error: Invalid operation. Please use +, -, *, /, //, %, or **."

print(f"\nResult: {num1} {operation} {num2} = {result}")
```

### Implementation of Assignment Operators

```python
def demonstrate_assignment_operators():
    print("\n--- Assignment Operators Demonstration ---")
    x = get_numeric_input("Enter an initial value for x: ")
    print(f"Initial x: {x}")

    value = get_numeric_input("Enter a value to use in operations (e.g., 5): ")

    x_original = x # Store original for reset

    print(f"\n-- Applying Compound Assignments --")
    x += value
    print(f"x += {value} (x = x + {value}): {x}")
    x = x_original # Reset for next operation

    x -= value
    print(f"x -= {value} (x = x - {value}): {x}")
    x = x_original

    x *= value
    print(f"x *= {value} (x = x * {value}): {x}")
    x = x_original

    if value != 0:
        x /= value
        print(f"x /= {value} (x = x / {value}): {x}")
        x = x_original

        x //= value
        print(f"x //= {value} (x = x // {value}): {x}")
        x = x_original

        x %= value
        print(f"x %= {value} (x = x % {value}): {x}")
        x = x_original
    else:
        print("Cannot perform division/modulus assignments with zero.")

    x **= value # Using original x for exponentiation
    print(f"x **= {value} (x = x ** {value}): {x}")
```

```python
def demonstrate_identity_operators():
    print("\n--- Identity Operators Demonstration ---")

    # Example with integers (Python often interns small integers)
    int_a = get_numeric_input("Enter a small integer for 'a': ", type=int)
    int_b = get_numeric_input("Enter another small integer for 'b': ", type=int)
    int_c = int_a # int_c refers to the same object as int_a

    print(f"\nIntegers:")
    print(f"a: {int_a}, id(a): {id(int_a)}")
    print(f"b: {int_b}, id(b): {id(int_b)}")
    print(f"c: {int_c}, id(c): {id(int_c)}")
    print(f"a is b: {int_a is int_b}")
    print(f"a is c: {int_a is int_c}")
    print(f"a is not b: {int_a is not int_b}")

    # Example with lists (usually different objects even with same content)
    list1_str = input("\nEnter elements for List 1 (comma-separated): ")
    list1 = [item.strip() for item in list1_str.split(',')]
    list2_str = input("Enter elements for List 2 (comma-separated): ")
    list2 = [item.strip() for item in list2_str.split(',')]
    list3 = list1 # list3 refers to the same object as list1

    print(f"\nLists:")
    print(f"List1: {list1}, id(List1): {id(list1)}")
    print(f"List2: {list2}, id(List2): {id(list2)}")
    print(f"List3: {list3}, id(List3): {id(list3)}")
    print(f"List1 is List2: {list1 is list2}")
    print(f"List1 is List3: {list1 is list3}")
    print(f"List1 is not List2: {list1 is not list2}")
```

### Implementation of Bitwise Operators

```python
def demonstrate_bitwise_operators():
    print("\n--- Bitwise Operators Demonstration ---")

    num1 = get_numeric_input("Enter the first integer (e.g., 10): ", type=int)
    num2 = get_numeric_input("Enter the second integer (e.g., 4): ", type=int)

    print(f"\nBinary representation: {num1} = {bin(num1)}, {num2} = {bin(num2)}")

    print(f"Bitwise AND ({num1} & {num2}): {num1 & num2} ({bin(num1 & num2)})")
    print(f"Bitwise OR ({num1} | {num2}): {num1 | num2} ({bin(num1 | num2)})")
    print(f"Bitwise XOR ({num1} ^ {num2}): {num1 ^ num2} ({bin(num1 ^ num2)})")
    print(f"Bitwise NOT (~{num1}): {~num1} ({bin(~num1)}) (Note: depends on system representation)")
    print(f"Left Shift ({num1} << 1): {num1 << 1} ({bin(num1 << 1)})")
    print(f"Right Shift ({num1} >> 1): {num1 >> 1} ({bin(num1 >> 1)})")

    # Example: Check if a number is even/odd using bitwise AND
    num_check = get_numeric_input("\nEnter an integer to check if it's even or odd: ", type=int)
    if (num_check & 1) == 0:
        print(f"{num_check} is Even.")
    else:
        print(f"{num_check} is Odd.")
```
