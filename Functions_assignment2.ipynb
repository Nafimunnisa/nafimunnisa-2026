## Lambda

```python
square_lambda = lambda x: x * x

# Demonstrate the lambda expression
number = 8
result = square_lambda(number)
print(f"The square of {number} is {result}")
```

```python
cube_lambda = lambda x: x * x * x

# Demonstrate the lambda expression
number_for_cube = 5
result_cube = cube_lambda(number_for_cube)
print(f"The cube of {number_for_cube} is {result_cube}")
```

```python
add_lambda = lambda a, b: a + b

# Demonstrate the lambda expression
a = 25
b = 17
sum_result = add_lambda(a, b)
print(f"The sum of {a} and {b} is {sum_result}")
```

```python
subtract_lambda = lambda a, b: a - b

# Demonstrate the lambda expression with the given values
a = 50
b = 18
subtract_result = subtract_lambda(a, b)
print(f"The result of subtracting {b} from {a} is {subtract_result}")
```

```python
multiply_lambda = lambda a, b: a * b

# Demonstrate the lambda expression with the given values
a = 12
b = 9
multiply_result = multiply_lambda(a, b)
print(f"The product of {a} and {b} is {multiply_result}")
```

```python
divide_lambda = lambda a, b: a / b

# Demonstrate the lambda expression with the given values
a = 100
b = 4
divide_result = divide_lambda(a, b)
print(f"The result of dividing {a} by {b} is {divide_result}")
```

```python
larger_lambda = lambda a, b: a if a > b else b

# Demonstrate the lambda expression with the given values
a = 72
b = 91
larger_number = larger_lambda(a, b)
print(f"The larger number between {a} and {b} is {larger_number}")
```

```python
smaller_lambda = lambda a, b: a if a < b else b

# Demonstrate the lambda expression with the given values
a = 46
b = 28
smaller_number = smaller_lambda(a, b)
print(f"The smaller number between {a} and {b} is {smaller_number}")
```

```python
is_even_lambda = lambda num: num % 2 == 0

# Demonstrate the lambda expression with the given number
number = 34
is_even = is_even_lambda(number)
print(f"Is {number} even? {is_even}")
```

```python
is_odd_lambda = lambda num: num % 2 != 0

# Demonstrate the lambda expression with the given number
number = 27
is_odd = is_odd_lambda(number)
print(f"Is {number} odd? {is_odd}")
```

```python
area_rectangle_lambda = lambda length, width: length * width

# Demonstrate the lambda expression with the given values
length = 12
width = 8
area = area_rectangle_lambda(length, width)
print(f"The area of a rectangle with length {length} and width {width} is {area}")
```

```python
area_circle_lambda = lambda radius: 3.14 * radius * radius

# Demonstrate the lambda expression with the given value
radius = 7
area = area_circle_lambda(radius)
print(f"The area of a circle with radius {radius} is {area}")
```

```python
perimeter_rectangle_lambda = lambda length, width: 2 * (length + width)

# Demonstrate the lambda expression with the given values
length = 15
width = 10
perimeter = perimeter_rectangle_lambda(length, width)
print(f"The perimeter of a rectangle with length {length} and width {width} is {perimeter}")
```

```python
celsius_to_fahrenheit_lambda = lambda celsius: (celsius * 9/5) + 32

# Demonstrate the lambda expression with the given value
celsius = 35
fahrenheit = celsius_to_fahrenheit_lambda(celsius)
print(f"{celsius}°C is equal to {fahrenheit}°F")
```

```python
fahrenheit_to_celsius_lambda = lambda fahrenheit: (fahrenheit - 32) * 5/9

# Demonstrate the lambda expression with the given value
fahrenheit = 104
celsius = fahrenheit_to_celsius_lambda(fahrenheit)
print(f"{fahrenheit}°F is equal to {celsius}°C")
```

```python
string_length_lambda = lambda s: len(s)

# Demonstrate the lambda expression with the given word
word = "Python"
length = string_length_lambda(word)
print(f"The length of the string '{word}' is {length}")
```

```python
first_char_lambda = lambda s: s[0]

# Demonstrate the lambda expression with the given word
word = "Azure"
first_character = first_char_lambda(word)
print(f"The first character of the string '{word}' is '{first_character}'")
```

```python
last_char_lambda = lambda s: s[-1]

# Demonstrate the lambda expression with the given word
word = "Databricks"
last_character = last_char_lambda(word)
print(f"The last character of the string '{word}' is '{last_character}'")
```

```python
absolute_value_lambda = lambda num: -num if num < 0 else num

# Demonstrate the lambda expression with the given number
number = -45
absolute_value = absolute_value_lambda(number)
print(f"The absolute value of {number} is {absolute_value}")
```

```python
grade_lambda = lambda marks: "Distinction" if marks >= 75 else ("Pass" if marks >= 40 else "Fail")

# Demonstrate the lambda expression with the given marks
marks = 72
grade = grade_lambda(marks)
print(f"For marks = {marks}, the grade is: {grade}")
```

```python
numbers = [1, 2, 3, 4, 5, 6]
squared_numbers = list(map(lambda x: x * x, numbers))
print(f"Original numbers: {numbers}")
print(f"Squared numbers: {squared_numbers}")
```
