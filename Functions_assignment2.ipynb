```python
from functools import reduce

numbers = [12, 45, 12, 67, 45, 89, 23, 67, 100]

# The accumulator 'acc' will be a tuple: (current_sum, seen_numbers_set)
# The initial value for acc is (0, set()), meaning sum is 0 and no numbers seen yet.
sum_of_unique_numbers, _ = reduce(lambda acc, num:
    (acc[0] + num, acc[1].union({num})) if num not in acc[1]
    else acc,
    numbers, (0, set()))

print(f"Original numbers: {numbers}")
print(f"Sum of unique numbers: {sum_of_unique_numbers}")
```

Section 5: Combined Lambda, map(), filter(), and reduce()

```python
numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

# Filter out even numbers
even_numbers = filter(lambda x: x % 2 == 0, numbers)

# Map to calculate the square of each even number
squared_even_numbers = list(map(lambda x: x * x, even_numbers))

print(f"Original numbers: {numbers}")
print(f"Squared even numbers: {squared_even_numbers}")
```

```python
from functools import reduce

numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

# Filter out odd numbers
odd_numbers = filter(lambda x: x % 2 != 0, numbers)

# Reduce to calculate the sum of odd numbers
sum_of_odd_numbers_filtered = reduce(lambda x, y: x + y, odd_numbers)

print(f"Original numbers: {numbers}")
print(f"Sum of odd numbers: {sum_of_odd_numbers_filtered}")
```

```python
numbers = [10, 25, 40, 55, 70, 85, 100]

# Filter numbers greater than 50
greater_than_50 = filter(lambda x: x > 50, numbers)

# Map to multiply every selected number by 2
multiplied_numbers = list(map(lambda x: x * 2, greater_than_50))

print(f"Original numbers: {numbers}")
print(f"Numbers greater than 50, multiplied by 2: {multiplied_numbers}")
```

```python
from functools import reduce

numbers = [-15, 20, -35, 40, 50, -10, 60]

# 1. Extract positive numbers using filter()
positive_numbers = filter(lambda x: x > 0, numbers)

# 2. Calculate the square of every positive number using map()
squared_positive_numbers = map(lambda x: x * x, positive_numbers)

# 3. Calculate the total sum of the squared values using reduce()
total_sum_of_squared_positives = reduce(lambda x, y: x + y, squared_positive_numbers)

print(f"Original numbers: {numbers}")
print(f"Total sum of squared positive numbers: {total_sum_of_squared_positives}")
```

```python
from functools import reduce

temperatures = [18, 24, 30, 35, 40, 12, 28, 33, 15, 42]

# 1. Extract temperatures greater than 30°C using filter()
temperatures_greater_than_30 = filter(lambda temp: temp > 30, temperatures)

# 2. Convert the selected values into Fahrenheit using map()
celsius_to_fahrenheit = lambda celsius: (celsius * 9/5) + 32
fahrenheit_temperatures = list(map(celsius_to_fahrenheit, temperatures_greater_than_30))

# 3. Calculate the average Fahrenheit temperature using reduce()
# First, calculate the sum
if fahrenheit_temperatures: # Check if the list is not empty to avoid error with reduce
    sum_fahrenheit = reduce(lambda x, y: x + y, fahrenheit_temperatures)
    average_fahrenheit = sum_fahrenheit / len(fahrenheit_temperatures)
else:
    average_fahrenheit = 0 # Or handle as appropriate for an empty list

print(f"Original temperatures (Celsius): {temperatures}")
print(f"Temperatures greater than 30°C: {list(filter(lambda temp: temp > 30, temperatures))}")
print(f"Converted Fahrenheit temperatures (from those > 30°C): {fahrenheit_temperatures}")
print(f"Average Fahrenheit temperature: {average_fahrenheit:.2f}°F")
```

```python
from functools import reduce

products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]

# 1. Extract products currently in stock using filter()
in_stock_products = filter(lambda product: product['stock'] > 0, products)

# 2. Calculate each product's inventory value (price × stock) using map()
product_inventory_values = list(map(lambda product: product['price'] * product['stock'], in_stock_products))

# 3. Calculate the total inventory value using reduce()
total_inventory_value_combined = reduce(lambda total, value: total + value, product_inventory_values, 0)

print(f"Original products: {products}")
print(f"Inventory values of in-stock products: {product_inventory_values}")
print(f"Total inventory value: {total_inventory_value_combined}")
```

```python
from functools import reduce

transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]

# 1. Extract transactions whose amount is greater than 5000 using filter()
high_value_transactions = filter(lambda t: t['amount'] > 5000, transactions)

# 2. Add 18% GST to every selected transaction amount using map()
transactions_with_gst = list(map(lambda t: t['amount'] * 1.18, high_value_transactions))

# 3. Calculate the grand total after GST using reduce()
grand_total_after_gst = reduce(lambda total, amount: total + amount, transactions_with_gst, 0)

print(f"Original transactions: {transactions}")
print(f"Transactions amounts after GST (for amounts > 5000): {transactions_with_gst}")
print(f"Grand total after GST: {grand_total_after_gst:.2f}")
```

```python
from functools import reduce

sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]

# 1. Extract active salespeople from the South region using filter()
south_active_salespeople = list(filter(lambda sp: sp['region'] == 'South' and sp['active'], sales_records))

# 2. Calculate a 5% incentive for each selected salesperson and create a structured list using map()
#    Each item in the list will be a dictionary with salesperson name, sales, and incentive.
incentive_details = list(map(lambda sp: {
    "name": sp['salesperson'],
    "sales": sp['sales'],
    "incentive": sp['sales'] * 0.05
}, south_active_salespeople))

# 3. Calculate the total incentive payable using reduce()
total_incentive_payable = reduce(lambda total, sp: total + sp['incentive'], incentive_details, 0)

print(f"Original sales records: {sales_records}")
print(f"Active salespeople from South region with incentive details: {incentive_details}")
print(f"Total incentive payable: {total_incentive_payable:.2f}")
```
